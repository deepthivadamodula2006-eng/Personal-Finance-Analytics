```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned dataset
df = pd.read_excel("../dataset/cleaned_dataset.xlsx")

# Display first 5 rows
df.head()
```

```python
#Monthly Income Distribution
plt.figure(figsize=(8,5))
sns.histplot(df["Monthly_Income"], bins=20, kde=True)

plt.title("Distribution of Monthly Income")
plt.xlabel("Monthly Income")
plt.ylabel("Number of Users")

plt.show()
```

```python
#Total Monthly Expenses Distribution
plt.figure(figsize=(8,5))
sns.histplot(df["Total_Monthly_Expenses"], bins=20, kde=True)

plt.title("Distribution of Total Monthly Expenses")
plt.xlabel("Total Monthly Expenses")
plt.ylabel("Number of Users")

plt.show()
```

```python
#Current Savings Distribution
plt.figure(figsize=(8,5))
sns.histplot(df["Current_Savings"], bins=20, kde=True)

plt.title("Distribution of Current Savings")
plt.xlabel("Current Savings")
plt.ylabel("Number of Users")

plt.show()
```

```python
#Financial Health Score Distribution
plt.figure(figsize=(8,5))
sns.histplot(df["Financial_Health_Score"], bins=20, kde=True)

plt.title("Financial Health Score Distribution")
plt.xlabel("Financial Health Score")
plt.ylabel("Number of Users")

plt.show()
```

```python
#Recommendation Distribution
plt.figure(figsize=(10,5))
sns.countplot(x="Recommendation", data=df)

plt.title("Recommendation Distribution")
plt.xticks(rotation=45)

plt.show()
```

```python
#Gender Distribution
plt.figure(figsize=(6,6))

df["Gender"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```

```python
#Occupation Distribution
plt.figure(figsize=(10,6))

sns.countplot(
    y="Occupation",
    data=df,
    order=df["Occupation"].value_counts().index
)

plt.title("Occupation Distribution")
plt.show()
```

```python
#City Distribution
plt.figure(figsize=(10,6))

sns.countplot(
    y="City",
    data=df,
    order=df["City"].value_counts().index
)

plt.title("City-wise User Distribution")
plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(14,8))

numeric_df = df.select_dtypes(include=['number'])

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
#Monthly Income vs Total Expenses
plt.figure(figsize=(8,5))

sns.scatterplot(
    x="Monthly_Income",
    y="Total_Monthly_Expenses",
    data=df
)

plt.title("Monthly Income vs Total Expenses")
plt.show()
```

```python
#Current Savings vs Monthly Income
plt.figure(figsize=(8,5))

sns.scatterplot(
    x="Monthly_Income",
    y="Current_Savings",
    data=df
)

plt.title("Current Savings vs Monthly Income")
plt.show()
```

```python
#Budget Utilization Distribution
plt.figure(figsize=(8,5))

sns.histplot(
    df["Budget_Utilization (%)"],
    bins=20,
    kde=True
)

plt.title("Budget Utilization Distribution")
plt.show()
```

```python
#Financial Health Score by Gender
plt.figure(figsize=(6,5))

sns.boxplot(
    x="Gender",
    y="Financial_Health_Score",
    data=df
)

plt.title("Financial Health Score by Gender")
plt.show()
```

```python
#Average Monthly Income by Occupation
plt.figure(figsize=(12,6))

avg_income = df.groupby("Occupation")["Monthly_Income"].mean().sort_values()

avg_income.plot(kind="barh")

plt.title("Average Monthly Income by Occupation")
plt.xlabel("Average Monthly Income")
plt.show()
```

```python
#Expense Category Analysis
expense_columns = [
    "Rent",
    "Food",
    "Transport",
    "Shopping",
    "Bills",
    "Entertainment",
    "EMI",
    "Other_Expenses"
]

expense_totals = df[expense_columns].sum()

plt.figure(figsize=(10,5))

expense_totals.plot(kind="bar")

plt.title("Total Spending by Expense Category")
plt.ylabel("Amount")
plt.xticks(rotation=45)

plt.show()
```
